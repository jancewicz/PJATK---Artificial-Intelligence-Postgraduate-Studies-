```python
import pandas as pd
import numpy as np
```

```python
data = {
  'animal': ['cat', 'cat', 'snake', 'dog', 'dog', 'cat', 'snake', 'cat', 'dog',
  'dog'],
  'name': ['Daisy', 'Bella', 'Noodle', 'Charlie', 'Max', 'Molly', 'Draco', 'Kenzo',
  'Milo', 'Cooper'],
  'age': [2.5, 3, 0.5, np.nan, 5, 2, 4.5, np.nan, 7, 3],
  'visits': [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
  'priority': ['yes', 'yes', 'no', 'yes', 'no', 'no', 'no', 'yes', 'no', 'no']
}

labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']
```

### Zadanie 1

```python
animals = pd.DataFrame(data, index=labels)
```

```python
animals
```

### Zadanie 2

```python
# Display first 3 rows
animals.head(3)
```

```python
# Dispay animal and age columns
animals[["animal", "age"]]
```

```python
# Data from indexes 3,4,8 from columns 'visits' and 'priority'
animals.iloc[[3, 4, 8]][['visits', 'priority']]
```

```python
# Rows where visits count is greater than 2
animals.loc[animals["visits"] > 2].dropna()
```

```python
# Rows where animals is a cat and and it's age is below 4 y.o
animals.loc[(animals["animal"] == "cat") & (animals["age"] < 4)]
```

```python
# Rows where animal age is is range 2-4 (including) y.o
age_range = animals["age"].between(2, 4, inclusive="both")
animals[age_range]
```

### Zadanie 3

```python
# Add new record to animals dataFrame
k = ["dog", "Buddy", 5.5, 2, False]
animals.loc["k"] = k
animals
```

```python
# Add 'price' column with value based on animal priority - 10 for non-priority, 20 for priority
animals["price"] = np.where(animals["priority"] == "yes", 20, 10)
animals
```

```python
animals["total"] = animals["price"] * animals["visits"]
animals
```

```python
# Delete priority column
animals.drop(columns=["priority"], axis=1, inplace=True)
```

```python
# Rename column name: animal to species
animals = animals.rename(columns={"animal": "species"})
```

### Zadanie 4

```python
# Reset index to integers
animals = animals.reset_index(drop=True)
animals
```

```python
# Set name column as df index
animals = animals.set_index("name")
animals
```

```python
# Drop row with "Max"
animals = animals.drop(index="Max")
animals
```

```python
# Reset index to integers, and keep 'name' as column
animals = animals.reset_index() 
```

```python
animals
```

### Zadanie 5

```python
animals["species"] = animals["species"].apply(str.capitalize)
```

```python
animals
```

```python
# Save dataframe to csv file without indexing
animals.to_csv("animals.csv", index=False)
```

### Zadanie 7

```python
cols_to_convert = ["RainToday", "RainTomorrow"]

# Use map to convert all "yes" values to True, otherwise insert False
weather[cols_to_convert] = weather[cols_to_convert].map(lambda x: x == "yes")
```

```python
weather
```

```python
wind_dir_cols_to_convert = ["WindGustDir", "WindDir9am", "WindDir3pm", ]
weather[wind_dir_cols_to_convert] = weather[wind_dir_cols_to_convert].astype("category")
```

```python
weather.info()
```

```python
date_mask = (weather["Date"] > "2008-2-1") & (weather["Date"] <= "2008-2-14")

weather.loc[date_mask]
```

```python
wind_mask = (weather["WindDir9am"] == "SSW") | (weather["WindDir9am"] == "SW")

weather.loc[wind_mask]
```
